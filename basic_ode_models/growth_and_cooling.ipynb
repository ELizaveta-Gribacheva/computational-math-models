```python
import numpy as np
import matplotlib.pyplot as plt
```

 модель експоненційного зростання 

$$\frac{dy}{dt} = r \cdot y$$

$$t_{i+1} = t_i + \Delta t$$

$$y_{i+1} = y_i + \Delta t \cdot \left(r \cdot y_i\right)$$

```python
r = 0.5  
y0 = 10  
T = 10 
dt = 0.1  

steps = int(T /dt)
y = np.zeros(steps + 1)
t = np.zeros(steps + 1)

y[0] = y0

for i in range(steps):
    t[i+1] = t[i] + dt
    dydt = r * y[i]
    y[i+1] = y[i] + dt * dydt

plt.figure(figsize=(8, 6))
plt.plot(t, y)
plt.xlabel('t')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

модель закону охолодження Ньютона

$$
\frac{dT}{dt} = -k(T - T_0)
$$

```python
T_a = 33  
T_0 = 88  
k = 0.1  
h = 0.1  
time_end = 60 
t_steps = np.arange(0, time_end, h)
n_steps = len(t_steps)
T = np.zeros(n_steps)
T[0] = T_0

for n in range(0, n_steps - 1):
    dT_dt = -k * (T[n] - T_a) 
    T[n + 1] = T[n] + h * dT_dt  

plt.figure(figsize=(10, 5))
plt.plot(t_steps, T)
plt.title("Newton's Law of Cooling")
plt.xlabel("Time")
plt.ylabel("Temperature")
plt.grid()
plt.show()
```

 модель аттрактора Лоренца

$$\frac{dx}{dt} = \sigma(y - x)$$
$$\frac{dy}{dt} = x(\rho - z) - y$$
$$\frac{dz}{dt} = xy - \beta z$$

```python
sigma = 12.0
rho = 40.0
beta = 8/3
dt = 0.01  
t_end = 100.0  
steps = int(t_end / dt)

x = np.full(steps, 1.0)
y = np.full(steps, 1.0)
z = np.full(steps, 1.0)

for i in range(steps - 1):
    
    dxdt = sigma * (y[i] - x[i])
    dydt = x[i] * (rho - z[i]) - y[i]
    dzdt = x[i] * y[i] - beta * z[i]
    
    x[i + 1] = x[i] + dxdt * dt
    y[i + 1] = y[i] + dydt * dt
    z[i + 1] = z[i] + dzdt * dt


fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, lw=0.5)
ax.set_title("Lorenz Attractor")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
plt.show()
```
