Програмно реалізувати модель експоненційного зростання населення

Для цього самостійно підібрати параметри для проведення
обчислювального експерименту. Здійснити моделювання для різних наборів
даних.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
def exponent(y0, a, steps):
    y_ttt = [y0]
    for t in range(steps):
        y_t = (1 + a) * y_ttt[-1]
        y_ttt.append(y_t)
    return y_ttt
```

```python
y0 = 100
const = 0.1 
consttime = 20
```

```python
results = exponent(y0, const, consttime)

plt.figure(figsize=(10, 6))
plt.plot(results, marker='o', linestyle='-', color='pink')

plt.show()
```

```python
def richard(z, t, a, b, m, n, g, h):
    x, y = z
    dxdt = a * y - m * x + g
    dydt = b * x - n * y + h
    return [dxdt, dydt]
```

```python
a, b = 1, 1    
m, n = 2, 2     
g, h = 10, 50   
```

Тут у нас є рівняння двох ліній виду:

0 = ay − mx + g,

Ця рівновага існує поки mn − ab ≠ 0.

```python
print (m * n - a * b)

t = np.linspace(0, 10, 100)
z0 = [9, 5]  

result_2 = odeint(richard, z0, t, args=(a, b, m, n, g, h))

plt.figure(figsize=(10, 6))
plt.plot(t, result_2[:, 0], label='Країна X')
plt.plot(t, result_2[:, 1], label='Країна Y')
plt.xlabel('Час')
plt.ylabel('Витрати')
plt.legend()
plt.grid(True)
plt.show()
```
